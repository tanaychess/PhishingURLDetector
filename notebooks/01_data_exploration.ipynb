# 01. Exploratory Data Analysis: Phishing URL Dataset**Author**: Tanay Samson Pathare (Bhonsala Military College)This notebook loads and analyzes the raw benchmark dataset by Vrbančič et al. (2020).

```python
import pandas as pdimport numpy as npimport matplotlib.pyplot as pltimport seaborn as snsimport yamlwith open("../config/config.yaml", "r") as f:    config = yaml.safe_load(f)df = pd.read_csv("../data/raw/dataset_small.csv")print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")df.head()
```

```python
# Class Balance Analysisphishing_counts = df["phishing"].value_counts()print(f"Class Distribution:{phishing_counts}")print(f"Phishing Percentage: {phishing_counts[1] / len(df) * 100:.2f}%")print(f"Legitimate Percentage: {phishing_counts[0] / len(df) * 100:.2f}%")plt.figure(figsize=(6, 4))sns.countplot(data=df, x="phishing", palette="Set2")plt.title("Class Balance (0: Legitimate, 1: Phishing)")plt.show()
```
